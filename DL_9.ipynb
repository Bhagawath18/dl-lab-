```python
import tensorflow as tf
import numpy as np

url = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"
file_path = tf.keras.utils.get_file("shakespeare.txt", url)

text = open(file_path, "r", encoding="utf-8").read().lower()
text = text[:20000]

chars = sorted(set(text))
c2i = {c:i for i,c in enumerate(chars)}
i2c = {i:c for i,c in enumerate(chars)}

seq_len = 40
X = []
y = []

for i in range(len(text)-seq_len):
    X.append([c2i[c] for c in text[i:i+seq_len]])
    y.append(c2i[text[i+seq_len]])

X = np.array(X)
y = np.array(y)

model = tf.keras.Sequential([
    tf.keras.layers.Embedding(len(chars),32),
    tf.keras.layers.LSTM(64),
    tf.keras.layers.Dense(len(chars),activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy"
)

model.fit(X,y,batch_size=128,epochs=1,verbose=1)

def generate(seed,length=200):
    seed = seed.lower()
    seq = [c2i.get(c,0) for c in seed]

    for _ in range(length):
        inp = np.array(seq[-seq_len:]).reshape(1,-1)
        pred = model.predict(inp,verbose=0)[0]
        next_idx = np.random.choice(len(pred),p=pred)
        seq.append(next_idx)

    return "".join(i2c[i] for i in seq)

seed = "shall i compare thee to a summer's day?\n"

print("\nSeed Text:")
print(seed)

print("\nGenerated Text:")
print(generate(seed,200))

print("\nText generation completed successfully.")
```
